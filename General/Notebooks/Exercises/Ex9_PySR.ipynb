### Problem

Galaxies consist of stars, planets, and vast clouds of gas and dust, all bound together by gravity. Most large galaxies have supermassive black holes at their centers, some with billions of times the Sun\'s mass. There are evidences that the black hole which inhabits a spiral galaxy influences the galaxy structure and dynamics. The structure of a spiral galaxy can be described considering the shape of its arms and the velocity ot its rotation.

Let us consider the simplified data presented in Table 1 of [Discovery of a Planar Black Hole Mass Scaling Relation for Spiral Galaxies](https://arxiv.org/pdf/2309.08986) in which researchers try to find e relation between these feature of spiral galaxies, namely
- mass $M$ (in dex solar masses) of the black hole which inhabits the galaxy, where $k\,  \text{dex} = 10^k$;
- [spiral-arm pitch angle](https://en.wikipedia.org/wiki/Pitch_angle_of_a_spiral) $\phi$ (in degrees);
- physical maximum velocity rotation $v$ (in $\frac{km}{s}$).

In the data set is also included the error (standard deviation) of the measurements.
